## Hypothesis testing

```python
stats.ttest_1samp(brain_size['VIQ'], 0)
```

```python
## 2 sample t-test for testing differences across populations
female_viq = brain_size[brain_size['Gender'] == 'Female']['VIQ']
male_viq = brain_size[brain_size['Gender'] == 'Male']['VIQ']
stats.ttest_ind(female_viq, male_viq) 
```

```python
## paired tests on repeated measurements for same individual
stats.ttest_ind(brain_size['FSIQ'], brain_size['PIQ'])   
```

```python
stats.ttest_rel(brain_size['FSIQ'], brain_size['PIQ'])
```

```python
stats.ttest_rel(brain_size['FSIQ'], brain_size['PIQ'], 0)
```

```python
stats.wilcoxon(brain_size['FSIQ'], brain_size['PIQ'])
```

#### Exercise 3: Test the difference between weights in males and females and use non parametric statistic to test the difference between VIQ in males and females

```python
## creating variables
males = brain_size[brain_size['Gender'] == 'Male']
females = brain_size[brain_size['Gender']=='Female']
```

```python
## conducting a t-test for weight
t, p = stats.ttest_ind(males['Weight'].dropna(), females['Weight'].dropna())
print(f"Weight t-test: t={t:.3f}, p={p:.3f}")
```

```python
## conducting a Mann-Whitney U for VIQ
u, p = stats.mannwhitneyu(males['VIQ'].dropna(), females['VIQ'].dropna())
print(f"VIQ Mann-Whitney: U={u:.3f}, p ={p:.3f}")
```

#### Exercise 4: test if the VIQ of male and female are different after removing the effect of brain size, height and weight.

```python
u, p = stats.mannwhitneyu(males['VIQ'].dropna(), females['VIQ'].dropna())
print(f"VIQ Mann-Whitney: U={u:.3f}, p={p:.3f}")
```

## Linear Models, multiple factors, and analysis of variance

```python
## generating simulated data according to the model
x = np.linspace(-5, 5, 20)
np.random.seed(1)

## adjusting for normal distributed noise
y = -5 + 3*x + 4 * np.random.normal(size = x.shape)

## creating a dataframe containing relevant variables
data = pd.DataFrame({'x':x, 'y':y})
```

```python
## specifying model (OLS) and fitting it
model = ols("y~x", data).fit()
```

```python
## getting the model summary
print(model.summary())
```

#### Exercise 5: Retrieve estimated parameters from the above model

```python
## printing estimated parameters
print(model.params)
```

#### Categorical variables

```python
## going back to brain size dataset and comparing IQ of male and female using linear model
model = ols("VIQ ~ C(Gender)", data = brain_size).fit()
```

```python
## printing paramters
print(model.params)
```

```python
## link to t-tests between different FSIQ and PIQ
data_fisq = pd.DataFrame({'iq': brain_size['FSIQ'], 'type': 'fsiq'})
```

```python
## creating piq dataframe
data_piq = pd.DataFrame({'iq': brain_size['PIQ'], 'type': 'piq'})
```

```python
## concat dataframes
data_long = pd.concat((data_fisq, data_piq))
data_long.head()
```

```python
data_long.tail()
```

```python
## fitting to a model
model = ols("iq ~ C(type)", data_long).fit()
```

```python
## printing summary results
print(model.summary())
```

```python
## statistical ttest on these dataframes
stats.ttest_ind(brain_size['FSIQ'], brain_size['PIQ'])
```

#### Multiple regression

```python
## reading iris.csv dataframe
iris_df = pd.read_csv('iris.csv')
iris_df.head()
```

```python
## fitting the model to the iris df
model = ols('Q("Sepal Width (cm)") ~ C(Species) + Q("Petal Length (cm)")', data=iris_df).fit()
print(model.params)
```

#### Post-hoc hypothesis testing: analysis of variance (ANOVA)

```python
print(model.f_test([0, 1, -1, 0]))
```

#### Exercise 6: Going back to the brain size + IQ data, test if the VIQ of male and female are different after removing the effect of brain size, height and weight.

```python
## creating a model to test this hypothesis
model = ols("VIQ ~ C(Gender) + MRI_Count + Height + Weight", data=brain_size).fit()
print(model.summary())
```
